```python
import numpy as np
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

X = np.array([[2, 8, 4, 5], [6, 3, 0, 8], [5, 4, 9, 1]])

# 1. 训练集fit标准化
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
print(X_scaled)
print('--' * 50)

# 此时 X_scaled 的每一列方差≈1
print(np.var(X_scaled,axis=0))
print('--' * 50)

# 2. 在标准化后的数据上做PCA
pca = PCA(n_components=0.9)
X_pca = pca.fit_transform(X_scaled)
print(X_pca)
print('-' * 50)
print(pca.components_)
print('--' * 50)

# PCA得到的主成分方差，不再等于1，从大到小排列
print(pca.explained_variance_)
print('--' * 50)
print(pca.explained_variance_ratio_)
```

```python
import numpy as np
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

X = np.array([[2, 8, 4, 5], [6, 3, 0, 8], [5, 4, 9, 1]])

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

pca = PCA(n_components=0.9)
X_pca = pca.fit_transform(X_scaled)

print("=== components_： 权重矩阵（特征向量）===")
print(pca.components_)
print("形状", pca.components_.shape)

print("\n=== explained_variance_ 特征值 ===")
print(pca.explained_variance_)
# 特征值大小：代表对应主成分携带信息量
```
